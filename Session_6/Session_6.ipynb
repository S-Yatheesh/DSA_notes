Searching and Counting

```python
text ="  HEY,HELLO ,Welcome to the Kerala IT Industry, show your skills here with many courses related to Data science and etc.  "

print(text.count("e"))     #  (Counts how many times 'e' appears)
print(text.find("Kerala")) #  (Returns the starting index position of the word)
print(text.find("Java"))   #  (Returns -1 because the word doesn't exist)
```

Changing Case

```python
text ="  HEY,HELLO ,Welcome to the Kerala IT Industry, show your skills here with many courses related to Data science and etc.  "
text_a ="  hEY,HELLO ,Welcome to the Kerala IT Industry, show your skills here with many courses related to Data science and etc.  "

print(text.upper())      # "WELCOME TO KERALA" (All uppercase)
print(text.lower())      # "welcome to kerala" (All lowercase)
print(text.title())      # "Welcome To Kerala" (Capitalizes every word)
print(text_a.strip().capitalize()) # "Welcome to kerala" (Capitalizes only the first letter)
print(text.swapcase()) #swap cases
```

Checking Conditions (Returns True or False)

```python
text = "HEY,HELLO ,Welcome to the Kerala IT Industry, show your skills here with many courses related to Data science and etc.  "

print(text.isalpha())    # False (Contains numbers, not just letters)
print(text.isalnum())    # True (Contains only letters and numbers)
print(text.startswith("HE")) # True (Checks the beginning)
print(text.endswith("  "))  # False (Checks the ending)
```
