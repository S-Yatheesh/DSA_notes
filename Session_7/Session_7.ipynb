Create a pgm where a user enter prices of 5 items into a list,Then apply following discount rules.
if total>2000 -> 20 % discount
elif total >1000 ->10% discount
else -> no discount
print the bill amount

```python
price_1=int(input("enter the amount of 1 st item :"))
price_2=int(input("enter the amount of 2 nd item :"))
price_3=int(input("enter the amount of 3 rd item :"))
price_4=int(input("enter the amount of 4 th item :"))
price_5=int(input("enter the amount of 5 th item :"))
total = int(price_1 + price_2+ price_3+ price_4+ price_5)
disc_above_20 = int(total)-(total *20)/100
disc_above_10= int(total)-(total *10)/100
if total >=2000 :
  print(f"Your bill amount is :{int(disc_above_20)}")
elif total>= 1000 < 2000 :
    print(f"Your bill amount is :{int(disc_above_10)}")
else :
    print(f"Your Bill amount is :{total}")
```
