BREAK,CONTINUE AND PASS

```python
for i in range(5):
  if i == 5:
    break
  print(i)
```

```python
for i in range(5):
  if i == 4:
    continue
  print(i)
```

```python
for i in range(5):
  if i == 3:
    pass
  print(i)
```

FUNCTIONS

A SET OF INSTRUCTIONS TO COMPLETE A PARTICUAR TASK WHICH CAN BE RECALLED FOR LATER PURPOSES.

SYNTAX :

           Def fname()
          ----------------------
          ----------------------
          ----------------------
          return

          fname() -< HERE the fuction can be recalled
          print("EOP")

```python
def greet():
  print("Hello,Gud mrngzzz")
print("Welcome ",end="")
greet()
greet()
greet()
print("EOP")
```

```python
def add():
  a=int(input("Enter the first number :"))
  b=int(input("Enter the second number :"))
  c=a+b
  print(f"Sum is {a} + {b} = {c}")
print("The Sum operation is deriving below : ")
add()
```

```python
def sub():
  a=int(input("Enter the first number :"))
  b=int(input("Enter the second number :"))
  c=a-b
  print(f" {a} - {b} = {c}")
print("The Subtraction operation is deriving below : ")
sub()
```

```python
def mul():
  a=int(input("Enter the first number :"))
  b=int(input("Enter the second number :"))
  c=a*b
  print(f" {a} * {b} = {c}")
print("The Product operation is deriving below : ")
mul()
```

```python
def div():
  a=int(input("Enter the first number :"))
  b=int(input("Enter the second number :"))
  c=a/b
  print(f" {a} / {b} = {c}")
print("The Division operation is deriving below : ")
div()
```
